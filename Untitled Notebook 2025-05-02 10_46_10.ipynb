```python
from pyspark.sql import SparkSession
from pyspark.sql import functions as F

# 1. Cria SparkSession
spark = SparkSession.builder \
    .appName("Pipeline CSV → Delta → PostgreSQL") \
    .getOrCreate()

# 2. Leitura dos CSVs
dfcomp = spark.read.csv(
    "/FileStore/tables/br_mdr_snis_municipio_agua_esgoto.csv",
    header=True,
    inferSchema=True
)
dfest = spark.read.csv(
    "/FileStore/tables/br_bd_diretorios_brasil_uf.csv",
    header=True,
    inferSchema=True
)
dfmun = spark.read.csv(
    "/FileStore/tables/br_bd_diretorios_brasil_municipio.csv",
    header=True,
    inferSchema=True
)

# 3. Padronização e tratamento de nulos
df = dfcomp \
    .withColumnRenamed("populacao_atentida_esgoto", "populacao_atendida_esgoto") \
    .na.fill(0, subset=[
        "populacao_urbana_residente_agua",
        "populacao_urbana_residente_esgoto"
    ])

# 4. Montagem da Dimensão Tempo
dim_tempo = df.select("ano").distinct()

# 5. Montagem da Dimensão Estado (UF)
dim_estado = df.select("sigla_uf") \
    .join(dfest, df.sigla_uf == dfest.sigla, "left") \
    .select(
        F.col("id_uf"),
        F.col("sigla").alias("sigla_uf"),
        F.col("nome").alias("estado_nome"),
        F.col("regiao")
    ).distinct()

# 6. Montagem da Dimensão Município
dim_municipio = dfmun.select(
    "id_municipio",
    "nome",
    "sigla_uf",
    "ddd",
    "nome_regiao",
    "amazonia_legal"
).distinct()

# 7. Construção da Tabela Fato
fato = df.alias("c") \
    .join(dim_estado.alias("e"),   "sigla_uf") \
    .join(dim_municipio.alias("m"), "id_municipio") \
    .select(
        "ano",
        "e.id_uf",
        "m.id_municipio",

        # Métricas de abastecimento e esgoto
        "populacao_atendida_agua",
        "populacao_atendida_esgoto",
        "extensao_rede_agua",
        "extensao_rede_esgoto",
        "volume_agua_produzido",
        "volume_esgoto_coletado",
        "volume_esgoto_tratado",
        # adicione aqui todas as colunas métricas que desejar...
    )

# 8. Escrita em Delta Lake
base_delta = "/mnt/delta/agua_esgoto"
dim_tempo.write.format("delta").mode("overwrite").save(f"{base_delta}/dim_tempo")
dim_estado.write.format("delta").mode("overwrite").save(f"{base_delta}/dim_estado")
dim_municipio.write.format("delta").mode("overwrite").save(f"{base_delta}/dim_municipio")
fato.write.format("delta") \
    .mode("overwrite") \
    .partitionBy("ano", "id_uf") \
    .save(f"{base_delta}/fato_abastecimento")

# 9. Configuração JDBC para PostgreSQL
jdbc_url = "jdbc:postgresql://127.0.0.1:5432/agua"
jdbc_props = {
    "user": "postgres",
    "password": "12501222",
    "driver": "org.postgresql.Driver",
    "numPartitions": "8",
    "batchsize": "5000"
}

# 10. Gravação no PostgreSQL
dim_tempo.write.jdbc(
    url=jdbc_url,
    table="dim_tempo",
    mode="overwrite",
    properties=jdbc_props
)
dim_estado.write.jdbc(
    url=jdbc_url,
    table="dim_estado",
    mode="overwrite",
    properties=jdbc_props
)
dim_municipio.write.jdbc(
    url=jdbc_url,
    table="dim_municipio",
    mode="overwrite",
    properties=jdbc_props
)
fato.write.jdbc(
    url=jdbc_url,
    table="fato_abastecimento",
    mode="overwrite",
    properties=jdbc_props
)

# 11. Opcional: Validação mínima
print("Count dim_tempo:", dim_tempo.count())
print("Count dim_estado:", dim_estado.count())
print("Count dim_municipio:", dim_municipio.count())
print("Count fato_abastecimento:", fato.count())

# Fim do pipeline
spark.stop()
```
